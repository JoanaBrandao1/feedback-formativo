```python
import os
import json
import time
import re

import sys
from pathlib import Path

import pandas as pd
from google import genai
from google.genai import types
from dotenv import load_dotenv
```

```python
load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    raise ValueError("GEMINI_API_KEY não encontrada no arquivo .env")

client = genai.Client(api_key=api_key)
```

```python
df = pd.read_csv("../data/raw/dataset.csv")
df.head()
```

```python
texto = df["Text"].iloc[0]
print(texto)
```

```python
MODEL = "gemini-2.5-flash-lite"
```

```python
generation_config = types.GenerateContentConfig(
    response_mime_type="application/json",
)
```

```python
prompt = f"""
Você é um especialista em avaliação educacional.

Sua tarefa é classificar o feedback abaixo de acordo com as categorias definidas.

Para cada categoria, responda com:
1 = presente no feedback
0 = não presente

Categorias:

GP1: Esclarece o que é um bom desempenho.
GP2: Promove autoavaliação e reflexão do aluno.
GP3: Fornece informações de qualidade sobre o aprendizado.
GP4: Incentiva diálogo entre aluno e professor/colegas.
GP5: Incentiva motivação e autoestima.
GP6: Ajuda a reduzir a diferença entre desempenho atual e desejado.
GP7: Fornece informações úteis para o professor ajustar o ensino.

Responda APENAS com um objeto JSON contendo as chaves GP1 a GP7, cada uma com valor 0 ou 1 (número inteiro).

Exemplo: {{"GP1": 0, "GP2": 1, "GP3": 0, "GP4": 0, "GP5": 0, "GP6": 1, "GP7": 0}}

Feedback:
{texto}
"""
```

```python
response = client.models.generate_content(
    model=MODEL,
    contents=prompt,
    config=generation_config,
)
print(response.text)
```

```python
resultado = json.loads(response.text)
resultado
```
